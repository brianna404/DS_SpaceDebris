## Model Training - Random Forest Classifier

```python
# import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
import ydf

# read data
debris_df = pd.read_csv("../Data/DS_combined_final.csv")
```

#### Model 1 - using all numerical features

```python
# remove non numerical columns
debris_df.drop(columns=['NORAD_CAT_ID','INTLDES', 'SATNAME', 'OBJECT_TYPE', 'COUNTRY', 'SITE', 'LAUNCH', 'DECAY'], inplace=True)
# split the data into training and test sets (80% training, 20% test)
train_df, test_df = train_test_split(debris_df, test_size=0.2, random_state=42)

print("Training set size:", len(train_df))
print("Test set size:", len(test_df))
```

```python
model = ydf.RandomForestLearner(label="LIFETIME_CLASS").train(train_df, verbose = 2) # label: defines the prediction column; train_ds: defines the training dataset, verbose: show training logs
```

```python
model.describe() # show model description
```

```python
# Evaluate the model
model.evaluate(test_df)
```

#### Model 2 - using all numerical features except REV_AT_EPOCH since it is basically a measure for the age of the object and it is highly correlated with the lifetime. That is beceause for every object the latest OMM file is used where the Epoch is likely very near the decay date.

```python
# remove REV_AT_EPOCH column
debris_df.drop(columns=['REV_AT_EPOCH'], inplace=True)

# split the data into training and test sets (80% training, 20% test)
train_df, test_df = train_test_split(debris_df, test_size=0.2, random_state=42)

print("Training set size:", len(train_df))
print("Test set size:", len(test_df))
```

```python
model = ydf.RandomForestLearner(label="LIFETIME_CLASS").train(train_df, verbose = 2) # label: defines the prediction column; train_ds: defines the training dataset, verbose: show training logs
```

```python
model.describe() # show model description
```

```python
# Evaluate the model
model.evaluate(test_df)
```
